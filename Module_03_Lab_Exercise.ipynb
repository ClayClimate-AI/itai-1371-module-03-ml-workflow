## Part 1: Understanding Types of Machine Learning

Machine learning can be categorized into three main types. Let's explore each with practical examples.

### 1. Supervised Learning
**Definition**: Learning from labeled examples to make predictions on new, unseen data.

**Examples**:
- **Classification**: Predicting categories (spam/not spam, disease/healthy)
- **Regression**: Predicting continuous values (house prices, temperature)

**Key Characteristic**: We have both input features (X) and correct answers (y) during training.

### 2. Unsupervised Learning
**Definition**: Finding hidden patterns in data without labeled examples.

**Examples**:
- **Clustering**: Grouping similar customers for marketing
- **Dimensionality Reduction**: Simplifying complex data while keeping important information

**Key Characteristic**: We only have input features (X), no correct answers during training.

### 3. Reinforcement Learning
**Definition**: Learning through trial and error by receiving rewards or penalties.

**Examples**:
- Game playing (chess, Go)
- Autonomous vehicles
- Recommendation systems that learn from user feedback

**Key Characteristic**: Agent learns by interacting with an environment and receiving feedback.

**For this course, we'll focus primarily on supervised learning, with some unsupervised learning in later modules.**

## Part 5: The Complete Machine Learning Workflow

Now let's implement the standard ML workflow step by step:

### The 6-Step ML Workflow:
1. **Data Preparation**: Clean and prepare the data
2. **Feature Selection**: Choose relevant input variables
3. **Data Splitting**: Separate training and testing data
4. **Model Training**: Teach the algorithm using training data
5. **Model Evaluation**: Test performance on unseen data
6. **Model Interpretation**: Understand what the model learned

Let's implement each step!

````python
# Step 1: Data Preparation
print("Step 1: Data Preparation")
print("=" * 30)

# Select features (X) and target (y)
# For simplicity, let's use the first 4 features
feature_names = ['alcohol', 'malic_acid', 'ash', 'alcalinity_of_ash']
X = df[feature_names]
y = df['wine_class']

# L4 assertions: shape + alignment (check_alignment) before anything downstream uses X/y
assert X.shape == (178, 4), f"unexpected feature matrix shape {X.shape}"
assert y.shape == (178,), f"unexpected target vector shape {y.shape}"
assert list(X.index) == list(y.index), "X and y are misaligned"

print(f"Selected features: {feature_names}")
print(f"Feature matrix shape: {X.shape}")
print(f"Target vector shape: {y.shape}")

# Display first few rows
print("\nFirst 5 samples:")
print(X.head())
````

````python
# Step 2: Data Splitting
print("Step 2: Data Splitting")
print("=" * 30)

# Split data into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2,      # 20% for testing
    random_state=42,    # For reproducible results
    stratify=y          # Maintain class proportions
)

# L4 assertions: split accounting, alignment, and stratification actually held
assert len(X_train) + len(X_test) == len(X), "train+test doesn't add up to full X"
assert len(X_train) == len(y_train) and len(X_test) == len(y_test), "X/y split misaligned"
assert set(y_train.unique()) == set(y_test.unique()) == set(y.unique()), "stratification failed -- a class is missing from a split"

print(f"Training set: {X_train.shape[0]} samples")
print(f"Testing set: {X_test.shape[0]} samples")
print(f"Training classes: {np.bincount(y_train)}")
print(f"Testing classes: {np.bincount(y_test)}")

print("\n🎯 Why split data?")
print("- Training set: Teach the model")
print("- Testing set: Evaluate performance on unseen data")
print("- This prevents overfitting (memorizing vs. learning)")
````

````python
# Step 3: Model Training
print("Step 3: Model Training")
print("=" * 30)

# Scale features: fit ONLY on training data, then apply to both sets.
# LogisticRegression converges reliably on standardized features.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# L4 assertions: scaling actually centered the data and produced no non-finite values
assert abs(X_train_scaled.mean()) < 0.01, "training data not properly centered after scaling"
assert np.isfinite(X_train_scaled).all() and np.isfinite(X_test_scaled).all(), "non-finite values after scaling"

# Create and train two different models
models = {
    'Logistic Regression': LogisticRegression(random_state=42),
    'Decision Tree': DecisionTreeClassifier(random_state=42, max_depth=3)
}

trained_models = {}

for name, model in models.items():
    print(f"\nTraining {name}...")

    # Train the model on scaled features
    model.fit(X_train_scaled, y_train)
    trained_models[name] = model

    print(f"✅ {name} training completed!")

# L4 assertion: both models actually trained
assert len(trained_models) == len(models) == 2, "not all models trained"

print("\n🤖 What happened during training?")
print("- Features were standardized (fit on train only, then applied to test)")
print("- Models learned patterns from training data")
print("- They found relationships between features and wine classes")
print("- Now they can make predictions on new data!")
````

````python
# Step 4: Model Evaluation
print("Step 4: Model Evaluation")
print("=" * 30)

results = {}
train_results = {}

for name, model in trained_models.items():
    # Make predictions on the scaled test set
    y_pred = model.predict(X_test_scaled)

    # Calculate accuracy
    accuracy = accuracy_score(y_test, y_pred)
    results[name] = accuracy

    # L4 assertions: prediction length and accuracy range sanity
    assert len(y_pred) == len(y_test), f"{name}: prediction length mismatch"
    assert 0.0 <= accuracy <= 1.0, f"{name}: accuracy out of range"

    print(f"\n{name} Results:")
    print(f"Accuracy: {accuracy:.3f} ({accuracy*100:.1f}%)")

    # Detailed classification report
    print("\nDetailed Performance:")
    print(classification_report(y_test, y_pred, target_names=wine_data.target_names))

    # Train accuracy: a lightweight overfitting check ahead of the full learning-curve cell
    train_pred = model.predict(X_train_scaled)
    train_accuracy = accuracy_score(y_train, train_pred)
    train_results[name] = train_accuracy
    assert 0.0 <= train_accuracy <= 1.0, f"{name}: train accuracy out of range"

    gap = train_accuracy - accuracy
    flag = "⚠️ possible overfitting" if gap > 0.10 else "✅ train/test gap small"
    print(f"Train: {train_accuracy:.3f} | Test: {accuracy:.3f} | Gap: {gap:.3f} -- {flag}")

# L4 assertion: nothing silently dropped between the three per-model dicts
assert set(results.keys()) == set(train_results.keys()) == set(trained_models.keys()), "results/train_results out of sync with trained_models"

# Compare models
print("\n📊 Model Comparison:")
for name, accuracy in results.items():
    print(f"{name}: {accuracy:.3f}")

best_model = max(results, key=results.get)
print(f"\n🏆 Best performing model: {best_model}")
````

````python
# Step 5: Model Interpretation
print("Step 5: Model Interpretation")
print("=" * 30)

# Visualize confusion matrix for the best model
best_model_obj = trained_models[best_model]
y_pred_best = best_model_obj.predict(X_test_scaled)

plt.figure(figsize=(8, 6))
cm = confusion_matrix(y_test, y_pred_best)

# L4 assertions: shape, accounting, and cross-check against Cell 13's recorded accuracy
assert cm.shape == (3, 3), f"unexpected confusion matrix shape {cm.shape}"
assert cm.sum() == len(y_test), "confusion matrix doesn't account for all test predictions"
assert np.isclose(np.trace(cm) / cm.sum(), results[best_model]), "confusion matrix accuracy doesn't match Cell 13's recorded accuracy"

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=wine_data.target_names,
            yticklabels=wine_data.target_names)
plt.title(f'Confusion Matrix - {best_model}')
plt.xlabel('Predicted Class')
plt.ylabel('True Class')
plt.show()

print(f"\n🔍 Interpreting the Confusion Matrix:")
print("- Diagonal values: Correct predictions")
print("- Off-diagonal values: Misclassifications")
print("- Perfect model would have all values on diagonal")
````

## Step 6: Learning Curves — Overfitting Check (Amendment)

**Why this cell exists:** A single test-set accuracy number (Cell 13) and a confusion matrix (Cell 14) don't fully answer whether a model generalized or just memorized. A learning curve settles that by comparing training and cross-validation accuracy across increasing training-set sizes, for both models. Added by C0 Amendment (2026-09-11) — see `docs/STEP0_Initial_Contract.md`.

````python
# Step 6: Learning Curves (Overfitting Check)
print("Step 6: Learning Curves (Overfitting Check)")
print("=" * 30)

from sklearn.model_selection import learning_curve

train_sizes = np.linspace(0.3, 1.0, 8)  # start at 30%: below this, cv=5 can starve the smallest class

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, (name, model) in zip(axes, trained_models.items()):
    sizes, train_scores, val_scores = learning_curve(
        model, X_train_scaled, y_train,
        train_sizes=train_sizes,
        cv=5,
        scoring='accuracy',
        random_state=42
    )

    train_mean = train_scores.mean(axis=1)
    val_mean = val_scores.mean(axis=1)

    # L4 assertions: output shapes align and accuracies stay in [0, 1]
    assert sizes.shape[0] == train_scores.shape[0] == val_scores.shape[0], f"{name}: mismatched learning-curve output shapes"
    assert np.all((train_mean >= 0) & (train_mean <= 1)) and np.all((val_mean >= 0) & (val_mean <= 1)), f"{name}: accuracy out of range in learning curve"

    ax.plot(sizes, train_mean, 'o-', color='blue', label='Training score')
    ax.plot(sizes, val_mean, 'o-', color='orange', label='Validation score')
    ax.set_title(f'Learning Curve - {name}')
    ax.set_xlabel('Training Set Size')
    ax.set_ylabel('Accuracy')
    ax.legend(loc='best')
    ax.set_ylim(0, 1.05)

    final_gap = train_mean[-1] - val_mean[-1]
    flag = "⚠️ possible overfitting" if final_gap > 0.10 else "✅ small gap -- good generalization"
    print(f"{name}: final training acc={train_mean[-1]:.3f}, final validation acc={val_mean[-1]:.3f}, gap={final_gap:.3f} -- {flag}")

plt.tight_layout()
plt.show()

print("\n📉 Interpreting Learning Curves:")
print("- Training line high + validation line low + big gap => overfitting (memorizing)")
print("- Both lines low and close together => underfitting")
print("- Both lines high and close together => healthy generalization")
````

### My Understanding of Machine Learning Types

**Supervised Learning**: Learning from labeled examples, meaning data where the correct answer is already known. The model can check its own predictions against the truth while it trains. The Wine dataset's three known cultivar labels are a good example.

**Why do you think this model performed better?**: Logistic Regression draws one clean, straight decision boundary between the three wine classes.
